# 01. 전신 reachability와 순기구학

## 목표
2링크 팔의 손 위치를 계산하고, 팔만으로 닿지 않는 목표에 몸체(base)를 이동하면 작업공간이 어떻게 넓어지는지 확인합니다. 실제 humanoid 제어기가 아닌 전신 제어의 기초 직관을 위한 toy model입니다.

```python
from math import cos, sin, radians, hypot

def hand_position(base_x, base_y, shoulder, elbow, lengths=(0.55, 0.45)):
    # 두 번째 링크의 각도는 첫 관절 각도에 팔꿈치 각도를 더합니다.
    a1, a2 = radians(shoulder), radians(shoulder + elbow)
    x = base_x + lengths[0] * cos(a1) + lengths[1] * cos(a2)
    y = base_y + lengths[0] * sin(a1) + lengths[1] * sin(a2)
    return round(x, 3), round(y, 3)

print('손 위치:', hand_position(0.0, 0.0, 35, -50))
```

```python
def reachable(base, target, arm_length=1.0, safety_margin=0.05):
    distance = hypot(target[0] - base[0], target[1] - base[1])
    return distance <= arm_length - safety_margin, distance

target = (1.35, 0.25)
for base in [(0.0, 0.0), (0.35, 0.0), (0.55, 0.1)]:
    ok, distance = reachable(base, target)
    print(f'base={base}, 거리={distance:.2f} m, 안전하게 도달={ok}')
# base 이동은 단순히 범위를 늘리는 것뿐 아니라 무게중심·충돌 제약도 바꿉니다.
```
