# 01. Retrieval과 ACL 기초
목표: 작은 corpus에서 lexical score를 계산하고 사용자 권한을 검색 단계에 적용합니다. Google Cloud API를 호출하지 않는 toy lab입니다.

```python
import re
docs=[
 {'id':'d1','text':'환불 요청 기한은 구매 후 7일입니다','roles':{'customer','support'}},
 {'id':'d2','text':'관리자 전용 보안 사고 대응 절차입니다','roles':{'admin'}},
 {'id':'d3','text':'배송 조회는 주문 번호가 필요합니다','roles':{'customer','support'}},
]
def tokens(text): return set(re.findall(r'[가-힣A-Za-z0-9]+',text.lower()))
def search(query,role,k=2):
 q=tokens(query); allowed=[d for d in docs if role in d['roles']]
 ranked=sorted(((len(q & tokens(d['text'])),d) for d in allowed),key=lambda item:(-item[0],item[1]['id']))
 return [d for score,d in ranked if score>0][:k]
result=search('환불 요청 기한은','customer')
print(result); assert [d['id'] for d in result]==['d1']; assert all('customer' in d['roles'] for d in result)
```

실제 시스템에서도 검색 후가 아니라 검색 전에 identity·ACL을 적용합니다. 검색 실패와 관련 문서 0건을 서로 다른 상태로 처리하세요.
