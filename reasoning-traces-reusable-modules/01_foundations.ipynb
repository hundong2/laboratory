# 01. 원자 skill과 routing 기초

목표: 같은 skill inventory라도 routing이 달라지면 trace와 결과가 달라짐을 확인합니다. Python 표준 라이브러리만 사용합니다.

```python
def reverse(text):
    return text[::-1]

def rotate_left(text):
    return text[1:] + text[:1] if text else text

def duplicate(text):
    return text + text

skills = {"reverse": reverse, "rotate": rotate_left, "duplicate": duplicate}
print({name: fn("abc") for name, fn in skills.items()})
```

## Routing을 분리해 표현하기

`previous`는 직전 출력을 사용하고, `original`은 원래 입력을 다시 사용합니다. Skill은 같아도 중간 정보의 전달 규칙이 다르면 다른 합성이 됩니다.

```python
def run_trace(original, plan):
    previous = original
    trace = []
    for skill_name, route in plan:
        source = previous if route == "previous" else original
        result = skills[skill_name](source)
        trace.append({"skill": skill_name, "route": route, "input": source, "output": result})
        previous = result
    return previous, trace

plan_a = [("reverse", "previous"), ("duplicate", "previous")]
plan_b = [("reverse", "previous"), ("duplicate", "original")]
for plan in (plan_a, plan_b):
    result, trace = run_trace("abc", plan)
    print(plan, "=>", result, trace)
```

## 조합 깊이와 OOD

훈련에서 깊이 1~2만 보았다고 가정하고 깊이 3 조합을 만듭니다. 모든 skill을 안다는 사실만으로 새 interface를 정확히 처리한다는 보장은 없습니다.

```python
seen_plans = [
    [("reverse", "previous")],
    [("rotate", "previous"), ("duplicate", "previous")],
]
ood_plan = [("reverse", "previous"), ("rotate", "original"), ("duplicate", "previous")]
result, trace = run_trace("abcd", ood_plan)
print("seen depths:", [len(plan) for plan in seen_plans])
print("OOD depth:", len(ood_plan), "result:", result)
for step in trace:
    print(step)
```
