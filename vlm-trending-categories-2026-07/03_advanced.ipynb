# 03. Advanced: 트렌드 분류 미니 파이프라인

## 목표

새로운 VLM 트렌드 문장이 들어왔을 때, 키워드 기반으로 카테고리를 추천하는 작은 파이프라인을 만듭니다. 실제 연구 자동화에서는 임베딩 검색이나 LLM 분류기를 쓰겠지만, 여기서는 규칙 기반으로 구조를 이해합니다.


```python
category_rules = {
    "document_memory": ["ocr", "document", "pdf", "kv", "cache", "scroll", "layout"],
    "unified_generation": ["generation", "segmentation", "depth", "pose", "diffusion", "caption"],
    "spatial_robust": ["boundary", "event", "robust", "misinformation", "spatial", "rgb"],
    "interactive_world_vla": ["world", "video", "robot", "vla", "simulation", "action", "agent"],
}

def classify_update(text):
    text_lower = text.lower()
    scores = {}
    for category, keywords in category_rules.items():
        scores[category] = sum(1 for keyword in keywords if keyword in text_lower)
    best = max(scores, key=scores.get)
    return best, scores

samples = [
    "A new OCR model keeps KV cache constant for long PDF parsing.",
    "A video diffusion backbone performs depth and pose estimation.",
    "An event camera augmented VLM improves robustness under low light.",
    "A robot VLA model learns actions in simulation-ready 3D worlds.",
]

for sample in samples:
    print(sample)
    print(classify_update(sample))
    print()
```

```python
learning_paths = {
    "document_memory": ["KV cache", "OCR pipeline", "layout-aware document RAG"],
    "unified_generation": ["output schema", "mask/depth serialization", "diffusion backbone"],
    "spatial_robust": ["boundary map", "event camera", "modality conflict benchmark"],
    "interactive_world_vla": ["world state", "action space", "sim-to-real evaluation"],
}

def recommend_path(update_text):
    category, scores = classify_update(update_text)
    return {"category": category, "scores": scores, "next_topics": learning_paths[category]}

print(recommend_path("New mobile GUI agent reads long screens with scroll trajectories."))
```

- 날짜별 upvote, GitHub star, arXiv 제출일을 따로 저장해 최신성과 인기도를 분리합니다.
- 카테고리별로 평가 질문을 자동 생성합니다. 예: 문서 VLM은 메모리, 공간 VLM은 modality conflict, VLA는 action latency를 묻습니다.
