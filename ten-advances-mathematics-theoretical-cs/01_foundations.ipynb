# 01. 점근 경계와 지수 개선

## 목표
점근 표기와 지수의 작은 변화가 큰 차원에서 얼마나 큰 비율 차이를 만드는지 계산합니다. 논문의 신규 증명을 재현하지 않고 sphere-packing exponent와 산술 formula 하한을 읽는 법에 집중합니다.

```python
from math import log2

def packing_upper_bound(dimension, exponent):
    """Return the toy exponential upper bound 2^(-exponent * d)."""
    return 2 ** (-exponent * dimension)

classical = 0.59905576
new = 0.6044
for dimension in (10, 100, 1_000):
    old_bound = packing_upper_bound(dimension, classical)
    new_bound = packing_upper_bound(dimension, new)
    print(dimension, f'new/old={new_bound / old_bound:.6g}')
# 상한 지수가 커지면 2^(-alpha*d)는 더 작아져 더 강한 상한입니다.
```

```python
def formula_lower_bound(n):
    return n**4 / log2(n)

def circuit_lower_bound(n):
    return n**2 * log2(log2(n))

print(' n | formula toy scale | circuit toy scale')
for n in (16, 64, 256, 1024):
    print(f'{n:4} | {formula_lower_bound(n):17.1f} | {circuit_lower_bound(n):17.1f}')
# Ω 표기는 숨은 상수가 있으므로 이 값이 실제 최소 gate/leaf 수는 아닙니다.
# 여기서는 증가 차수만 비교합니다.
```
