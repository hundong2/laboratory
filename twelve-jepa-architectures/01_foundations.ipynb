# 01. JEPA 유형 분류 기초

목표: 12개 JEPA 연구를 모달리티, 구조, downstream 목표로 분류한다. 위에서 아래로 실행하며 `jepa_types`의 조건을 바꿔 보자. 외부 패키지는 필요 없다.

```python
# 각 항목은 논문의 모든 세부 사항이 아니라 비교에 필요한 핵심 축만 담는다.
jepa_types = [
    {"name": "I-JEPA", "modality": "image", "structure": "space", "goal": "semantic content"},
    {"name": "MC-JEPA", "modality": "video", "structure": "space+time", "goal": "content+motion"},
    {"name": "V-JEPA", "modality": "video", "structure": "space+time", "goal": "motion+appearance"},
    {"name": "UI-JEPA", "modality": "UI sequence", "structure": "time", "goal": "user intent"},
    {"name": "A-JEPA", "modality": "audio", "structure": "time+frequency", "goal": "audio semantics"},
    {"name": "S-JEPA", "modality": "EEG", "structure": "channel+time", "goal": "cross-dataset transfer"},
    {"name": "TI-JEPA", "modality": "text+image", "structure": "cross-modal", "goal": "alignment+transfer"},
    {"name": "T-JEPA", "modality": "table", "structure": "feature subset", "goal": "tabular representation"},
    {"name": "ACT-JEPA", "modality": "observation+action", "structure": "time+action", "goal": "policy representation"},
    {"name": "Brain-JEPA", "modality": "fMRI", "structure": "ROI+time", "goal": "brain dynamics"},
    {"name": "3D-JEPA", "modality": "3D scene", "structure": "3D space", "goal": "3D representation"},
    {"name": "Point-JEPA", "modality": "point cloud", "structure": "point patches", "goal": "point representation"},
]
len(jepa_types)
```

```python
from collections import Counter

# 모달리티별 연구 수를 세면 목록이 어디에 집중됐는지 빠르게 볼 수 있다.
modality_counts = Counter(item["modality"] for item in jepa_types)
for modality, count in sorted(modality_counts.items()):
    print(f"{modality:>18}: {count}")
```

```python
def find_by_structure(keyword):
    """구조 설명에 keyword가 들어간 연구 이름을 반환한다."""
    return [
        item["name"]
        for item in jepa_types
        if keyword.lower() in item["structure"].lower()
    ]

print("시간 구조를 쓰는 연구:", find_by_structure("time"))
print("공간 구조를 쓰는 연구:", find_by_structure("space"))
```
