# 인터랙티브 월드/VLA 실습

## 목표

- world model의 상태-행동-관찰 루프를 구현합니다.
- pilot agent와 director agent의 역할을 분리합니다.
- VLA action space 확장 체크리스트를 만듭니다.


```python
state = {"room": 0, "energy": 3, "objects": ["door", "cube"], "log": []}

def pilot_agent(s):
    if s["energy"] <= 0:
        return "rest"
    if "cube" in s["objects"]:
        return "pick_cube"
    return "move"

def director_agent(s, action):
    if action == "pick_cube":
        return "a new target appears behind the cube"
    if action == "move":
        return "the next room becomes visible"
    return "lighting stabilizes"

def step(s):
    s = {k: (v[:] if isinstance(v, list) else v) for k, v in s.items()}
    action = pilot_agent(s)
    event = director_agent(s, action)
    if action == "pick_cube":
        s["objects"].remove("cube")
        s["energy"] -= 1
    elif action == "move":
        s["room"] += 1
        s["energy"] -= 1
    else:
        s["energy"] += 1
    s["log"].append({"action": action, "event": event})
    return s

for _ in range(5):
    state = step(state)
    print(state)
```

```python
action_spaces = {
    "single_arm": ["x", "y", "z", "gripper"],
    "mobile_manipulator": ["base_x", "base_y", "yaw", "arm_x", "arm_y", "arm_z", "gripper"],
    "whole_body": ["head", "waist", "base", "left_arm", "right_arm", "hands"],
}

for name, dims in action_spaces.items():
    print(name, "degrees:", len(dims), dims)
```
