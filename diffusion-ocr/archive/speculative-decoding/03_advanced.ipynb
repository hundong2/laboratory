# 03 · 이론 속도 곡면

논문의 이상화 식으로 승인률, 초안 비용, 초안 길이의 상호작용을 탐색한다. 실제 장치 throughput 예측기가 아니다.

**학습 목표**: $\alpha$, $c$, $\gamma$를 바꾸며 이상화 속도식의 최적 초안 길이와 민감도를 해석한다.

**실행 방법**: Python 3/Jupyter에서 셀을 위에서 아래 순서로 실행한다. 외부 패키지는 없으며 Python 표준 기능만 사용한다.

```python
# alpha=1은 일반 분수식이 0/0이 되므로 극한값을 명시적으로 분기한다.
def expected_tokens(alpha, gamma):
    if alpha == 1.0:
        return gamma + 1
    return (1 - alpha ** (gamma + 1)) / (1 - alpha)

def ideal_speedup(alpha, gamma, cost_ratio):
    return expected_tokens(alpha, gamma) / (1 + gamma * cost_ratio)
```

```python
for cost_ratio in (0.01, 0.05, 0.20):
    print(f'\nc={cost_ratio:.2f}')
    for alpha in (0.50, 0.70, 0.90):
        options = [(ideal_speedup(alpha, gamma, cost_ratio), gamma) for gamma in range(1, 11)]
        speed, gamma = max(options)
        print(f'alpha={alpha:.2f}: best gamma={gamma:2d}, ideal={speed:.3f}x')
        assert speed > 0
```

```python
alpha, gamma = 0.8, 4
for c in (0.00, 0.05, 0.10, 0.25, 0.50):
    print(f'c={c:.2f} -> {ideal_speedup(alpha, gamma, c):.3f}x')
assert ideal_speedup(alpha, gamma, 0.05) > ideal_speedup(alpha, gamma, 0.25)
```

$\gamma$의 최적값은 승인률이 아니라 초안 비용 $c$와 함께 결정된다. production에서는 목표 병렬 검증이 한 token forward와 같은 비용이라는 가정도 직접 측정해야 한다.
