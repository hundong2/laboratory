# 02. 작은 이진 부호 완전 탐색

## 목표
길이 `n`인 bit string 가운데 모든 쌍의 Hamming distance가 최소 `d` 이상인 가장 큰 집합을 작은 크기에서 완전 탐색합니다. 이는 논문의 지수적 상한을 재현하지 않는 toy model입니다.

```python
from itertools import combinations

def hamming_distance(left, right):
    return sum(a != b for a, b in zip(left, right))

def words(length):
    return [format(value, f'0{length}b') for value in range(2**length)]

print(hamming_distance('0000', '1011'))
```

```python
def maximum_code(length, minimum_distance):
    candidates = words(length)
    compatible = {
        word: {other for other in candidates if hamming_distance(word, other) >= minimum_distance}
        for word in candidates
    }
    best = []

    def search(chosen, remaining):
        nonlocal best
        if len(chosen) + len(remaining) <= len(best):
            return
        if not remaining:
            if len(chosen) > len(best):
                best = chosen
            return
        pivot = remaining[0]
        search(chosen + [pivot], [w for w in remaining[1:] if w in compatible[pivot]])
        search(chosen, remaining[1:])

    search([], candidates)
    return best

for distance in (1, 2, 3, 4):
    code = maximum_code(4, distance)
    print(f'A(4,{distance}) = {len(code)}: {code}')
```

```python
def verify_code(code, minimum_distance):
    return all(
        hamming_distance(left, right) >= minimum_distance
        for left, right in combinations(code, 2)
    )

candidate = maximum_code(5, 3)
print('candidate size:', len(candidate))
print('certificate valid:', verify_code(candidate, 3))
# verifier는 주어진 code가 유효하다는 하한 certificate를 확인합니다.
# maximum임을 보이려면 더 큰 code가 없다는 별도의 상한 논증이 필요합니다.
```
