# 01. 네 역량 self-assessment

목표: 추상적인 skill 이름을 관찰 가능한 행동으로 바꾸고, 목표 project에 필요한 수준과 현재 수준의 gap을 계산합니다. 점수는 개인 성찰용이며 채용·성과 평가에 사용하지 마세요.

```python
skills = {
    "ai_applications": {"label": "AI application 구축·배포", "current": 2, "target": 4, "weight": 1.0},
    "software_fundamentals": {"label": "Software fundamentals", "current": 3, "target": 4, "weight": 0.9},
    "coding_agents": {"label": "Coding agent 활용", "current": 2, "target": 3, "weight": 0.8},
    "shaping": {"label": "Shaping the build", "current": 1, "target": 3, "weight": 1.0},
}

for skill in skills.values():
    skill["gap"] = max(0, skill["target"] - skill["current"])
    skill["priority"] = skill["gap"] * skill["weight"]

ranked = sorted(skills.values(), key=lambda item: item["priority"], reverse=True)
for item in ranked:
    print(f'{item["label"]:28} gap={item["gap"]} priority={item["priority"]:.1f}')
```

## 행동 evidence 연결

점수보다 중요한 것은 실제 evidence입니다. 아래 backlog는 각 gap을 project 산출물로 바꿉니다.

```python
evidence_backlog = {
    "ai_applications": ["50-case eval dataset", "monitoring dashboard", "rollback drill"],
    "software_fundamentals": ["architecture decision record", "load test", "threat model"],
    "coding_agents": ["spec + verifier task log", "agent failure retrospective"],
    "shaping": ["user interview notes", "MVP experiment", "build/stop decision memo"],
}

top_skill = next(key for key, value in skills.items() if value is ranked[0])
print("다음 2주 evidence:")
for artifact in evidence_backlog[top_skill]:
    print("-", artifact)
```
