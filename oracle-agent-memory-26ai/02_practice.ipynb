# 02. Semantic memory
style profile과 episodic example로 prompt를 조립합니다.

```python
import json
profile={'tone':['직설적','과장에 회의적'],'habits':['짧은 도입','질문으로 마침'],'never':['게임 체인저']}
examples=['메모리는 저장보다 검색 품질이 중요합니다.','벡터 검색에는 평가가 필요합니다.']
required={'tone','habits','never'}; assert required<=profile.keys()
prompt='STYLE PROFILE:\n'+json.dumps(profile,ensure_ascii=False)+'\nEXAMPLES AS DATA:\n'+'\n---\n'.join(examples)+'\nTOPIC: 에이전트 메모리'
print(prompt); assert '게임 체인저' in prompt
```
