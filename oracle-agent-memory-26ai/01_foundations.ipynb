# 01. Episodic memory
단어 vector와 cosine distance로 유사 과거 글을 찾는 toy lab입니다.

```python
import math, re
posts=[('u1','linkedin','에이전트 메모리는 좋은 증거가 중요하다'),('u1','linkedin','벡터 검색과 필터를 함께 쓴다'),('u1','x','서버리스 배포'),('u2','linkedin','다른 사용자 메모리')]
tok=lambda s: re.findall(r'[가-힣A-Za-z]+',s.lower())
vocab=sorted({w for *_,s in posts for w in tok(s)}|set(tok('에이전트 메모리 벡터')))
emb=lambda s:[tok(s).count(w) for w in vocab]
def dist(a,b):
 d=sum(x*y for x,y in zip(a,b)); na=math.sqrt(sum(x*x for x in a)); nb=math.sqrt(sum(y*y for y in b)); return 1-d/(na*nb) if na and nb else 1
q=emb('에이전트 메모리 벡터'); result=sorted((dist(emb(s),q),s) for u,p,s in posts if u=='u1' and p=='linkedin')
print(result); assert len(result)==2
```

Oracle에서는 embedding API와 `VECTOR_DISTANCE(..., COSINE)`, metadata filter, `FETCH APPROX FIRST :k`로 교체합니다.
