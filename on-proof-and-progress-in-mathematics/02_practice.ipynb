# 02. 독자별 증명 설명 설계

목표: 같은 논증도 청중의 mental infrastructure에 따라 필요한 설명량이 달라진다는 주장을 정보 층 설계로 연습한다.

## 사례: 홀수 두 개의 합은 짝수다

전문가에게는 한 줄이면 충분하지만 초보자에게는 홀수·짝수의 정의와 변수의 역할까지 필요하다.

```python
layers = {
    "claim": "홀수 두 개의 합은 짝수다.",
    "definitions": "홀수는 2k+1, 짝수는 2k 꼴의 정수다.",
    "setup": "홀수 a=2m+1, b=2n+1을 잡는다.",
    "derivation": "a+b=2m+1+2n+1=2(m+n+1)이다.",
    "conclusion": "m+n+1은 정수이므로 합은 짝수다.",
    "intuition": "각 홀수에서 남는 하나씩을 합치면 새 쌍 하나가 된다.",
}

audience_profiles = {
    "expert": ["claim", "derivation"],
    "undergraduate": ["claim", "setup", "derivation", "conclusion"],
    "beginner": list(layers),
}


def render_explanation(profile: str) -> str:
    return "\n".join(layers[name] for name in audience_profiles[profile])


for profile in audience_profiles:
    text = render_explanation(profile)
    print(f"--- {profile} ({len(text)} chars) ---\n{text}\n")

assert len(render_explanation("beginner")) > len(render_explanation("expert"))
```

## 설명 품질을 확인하는 질문

길이가 길다고 좋은 설명은 아니다. 독자가 핵심 구조를 재구성할 수 있는지 확인해야 한다.

```python
questions = {
    "definition": "홀수를 왜 2k+1로 쓸 수 있는가?",
    "variable": "m과 n이 서로 달라도 되는 이유는 무엇인가?",
    "key_step": "합에서 2를 어떻게 묶었는가?",
    "transfer": "홀수 세 개의 합은 홀수인가 짝수인가?",
}

required_by_profile = {
    "expert": {"key_step"},
    "undergraduate": {"variable", "key_step", "transfer"},
    "beginner": set(questions),
}

for profile, required in required_by_profile.items():
    print(profile, [questions[key] for key in sorted(required)])

assert required_by_profile["expert"] < required_by_profile["beginner"]
```

## 실습 과제

피타고라스 정리 또는 귀류법 한 개를 선택한다. `claim`, `definitions`, `visual model`, `key inference`, `transfer question` 층을 만들고 서로 다른 독자 세 명에게 필요한 층을 고른다. 독자가 스스로 다시 설명한 결과를 기록해 profile을 수정한다.
