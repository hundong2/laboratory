# 02. Claim과 Citation 연결
목표: 검색 단계의 source ID를 보존하고 답변 claim마다 근거를 명시합니다.

```python
sources={
 'policy-7d':{'title':'환불 정책','url':'https://example.invalid/policy','text':'구매 후 7일 안에 환불을 요청할 수 있습니다.'},
 'shipping':{'title':'배송 안내','url':'https://example.invalid/shipping','text':'배송 조회에는 주문 번호가 필요합니다.'},
}
answer={'claims':[
 {'text':'환불 요청 기한은 구매 후 7일입니다.','source_ids':['policy-7d']},
 {'text':'배송 조회에는 주문 번호가 필요합니다.','source_ids':['shipping']},
]}
def validate_citations(answer,sources):
 errors=[]
 for i,claim in enumerate(answer['claims']):
  if not claim['source_ids']: errors.append(f'claim {i}: citation 없음')
  for source_id in claim['source_ids']:
   if source_id not in sources: errors.append(f'claim {i}: 알 수 없는 source {source_id}')
 return errors
errors=validate_citations(answer,sources); print(errors); assert errors==[]
```

```python
rendered=[]
for claim in answer['claims']:
 labels=', '.join(f"[{sources[s]['title']}]" for s in claim['source_ids'])
 rendered.append(f"{claim['text']} {labels}")
print('\n'.join(rendered))
```

Production에서는 단순 문서 단위가 아니라 passage 위치와 version을 저장하고, 사용자가 열 수 없는 URL은 citation으로 노출하지 않습니다.
