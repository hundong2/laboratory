# 03. Reflective memory
근거가 있는 작은 diff만 적용하고 snapshot으로 rollback합니다.

```python
from copy import deepcopy
state={'version':1,'profile':{'tone':['직설적'],'never':['과장']}}; history=[]; allowed={'tone','never'}
def apply_diff(state,diff):
 assert len(diff['additions'])+len(diff['removals'])<=3
 before=deepcopy(state); p=deepcopy(state['profile'])
 for x in diff['additions']:
  assert x['field'] in allowed and x['evidence']; p.setdefault(x['field'],[]).append(x['value'])
 after={'version':state['version']+1,'profile':p}; history.append({'before':before,'diff':diff,'after':deepcopy(after)}); return after
diff={'additions':[{'field':'tone','value':'실용적','evidence':'최근 5개 중 4개에 절차 포함'}],'removals':[]}
state=apply_diff(state,diff); assert state['version']==2
state=deepcopy(history[-1]['before']); print(state); assert state['version']==1
```

실제 시스템은 transaction, optimistic version, idempotent job과 model·prompt version audit을 추가해야 합니다.
