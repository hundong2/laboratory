## 카테고리별 개수

피드 순위는 날짜마다 바뀌지만, 반복 등장하는 문제 축은 안정적입니다.


```python
from collections import Counter

counts = Counter(item["category"] for item in papers)
for category, count in counts.most_common():
    print(f"{category}: {count}")
```

## 날짜별 피드 출현 빈도

사용자 입력의 날짜별 Top 5를 간단히 재구성해, 어떤 논문이 여러 날 반복됐는지 확인합니다.


```python
daily_feeds = {
    "2026-07-12": ["VLV Auto-Encoder", "Unlimited OCR Works", "Constructive Apraxia", "Do Images Speak Louder than Words?", "SCoPE VLM"],
    "2026-07-13": ["Unlimited OCR Works", "Vidu S1", "EmbodiedGen V2", "Dense Spatial Perception", "LingBot-VLA 2.0"],
    "2026-07-14": ["Vidu S1", "GenCeption", "Dense Spatial Perception", "Unlimited OCR Works", "VLV Auto-Encoder"],
    "2026-07-15": ["Vidu S1", "Unlimited OCR Works", "Dense Spatial Perception", "Infinite Worlds", "RE-VLM"],
    "2026-07-16": ["Unlimited OCR Works", "Scalable Visual Pretraining", "SenseNova-Vision", "Dense Spatial Perception", "Infinite Worlds"],
}

appearances = Counter(title for titles in daily_feeds.values() for title in titles)
for title, count in appearances.most_common():
    print(f"{title:<38} {count} day(s)")
```
