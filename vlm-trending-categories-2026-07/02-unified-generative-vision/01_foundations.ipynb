# 통합 생성형 비전 실습

## 목표

- detection, depth, pose 출력을 하나의 생성 포맷으로 바꿉니다.
- 생성 결과가 스키마를 만족하는지 검사합니다.
- 태스크별 헤드 방식과 통합 생성 방식을 비교합니다.


```python
import json

def generate_detection(label, bbox):
    return json.dumps({"task": "detect", "objects": [{"label": label, "bbox_xyxy": bbox}]})

def generate_depth(depth_grid):
    return json.dumps({"task": "depth", "map": depth_grid})

def generate_pose(t, r):
    return json.dumps({"task": "pose", "translation": t, "rotation": r})

print(generate_detection("robot", [10, 20, 80, 140]))
print(generate_depth([[1.2, 1.4], [0.9, 1.0]]))
print(generate_pose([0, 0, 1], [0, 0, 0, 1]))
```

```python
required = {
    "detect": ["objects"],
    "depth": ["map"],
    "pose": ["translation", "rotation"],
}

def validate(text):
    data = json.loads(text)
    task = data.get("task")
    if task not in required:
        return False
    return all(key in data for key in required[task])

outputs = [
    generate_detection("table", [1, 2, 3, 4]),
    generate_depth([[1.0]]),
    '{"task":"pose","translation":[0,0,1]}',
]
for output in outputs:
    print(output, validate(output))
```

```python
comparison = [
    ("task-specific heads", "high per-task optimization", "many architectures"),
    ("unified generation", "one interface", "schema errors need validators"),
    ("video generative pretraining", "strong spatiotemporal prior", "fair evaluation is hard"),
]
for row in comparison:
    print(f"{row[0]} | strength={row[1]} | risk={row[2]}")
```
