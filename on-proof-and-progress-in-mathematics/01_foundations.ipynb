# 01. 한 개념의 여러 표상

목표: Thurston이 미분을 예로 든 이유를 계산으로 확인한다. 기호식, 변화율, 접선, 선형 근사와 확대 관점이 같은 국소 구조를 설명한다는 것을 본다. Python 표준 라이브러리만 사용한다.

## 1. 함수와 기호적 미분

함수 `f(x)=x²`와 점 `x₀=3`을 사용한다. 기호 규칙으로 얻은 미분값은 `2x₀=6`이다.

```python
def f(x: float) -> float:
    return x ** 2


x0 = 3.0
symbolic_derivative = 2 * x0
print("기호 규칙으로 얻은 미분값:", symbolic_derivative)
assert symbolic_derivative == 6.0
```

## 2. 변화율과 확대 관점

간격 `h`를 줄이며 할선의 기울기가 접선 기울기에 가까워지는지 본다. 확대경의 배율을 높이는 것과 같은 계산이다.

```python
def difference_quotient(x: float, h: float) -> float:
    return (f(x + h) - f(x)) / h


rows = []
for h in (1.0, 0.1, 0.01, 0.001):
    estimate = difference_quotient(x0, h)
    rows.append((h, estimate, abs(estimate - symbolic_derivative)))

for h, estimate, error in rows:
    print(f"h={h:>5}: slope={estimate:.6f}, error={error:.6f}")

assert rows[-1][2] < rows[0][2]
```

## 3. 최선의 선형 근사

점 `x₀` 부근에서 `f(x)`를 접선 `L(x)=f(x₀)+f'(x₀)(x-x₀)`로 바꿔 본다. 거리가 절반이 되면 이차 함수의 근사 오차는 대략 1/4이 된다.

```python
def tangent_line(x: float) -> float:
    return f(x0) + symbolic_derivative * (x - x0)


errors = []
for distance in (1.0, 0.5, 0.25, 0.125):
    x = x0 + distance
    error = abs(f(x) - tangent_line(x))
    errors.append(error)
    print(f"distance={distance:.3f}, approximation error={error:.6f}")

ratios = [errors[i + 1] / errors[i] for i in range(len(errors) - 1)]
assert all(abs(ratio - 0.25) < 1e-12 for ratio in ratios)
```
